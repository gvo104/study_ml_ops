## 1. Модель

* **Архитектура**:
  Нужно окончательно выбрать:

  * TF-IDF + LogisticRegression (простой вариант, быстро, для MVP)

* **Feature extraction**: TF-IDF / эмбеддинги
* **Confidence**: как считать для разных моделей (логистическая регрессия → predict_proba, BERT → softmax)

* **Порог PSI**: 0.2 принято, но можно уточнить для проекта
* **Frequency**: каждые 100 запросов / раз в день

## 6. Переобучение

* **Стратегия формирования train_new**:

  * Sliding window (последние N новых + часть старого train)
  * Proportion: 80% confident + agree / 20% случайных старых или production
* **Обработка несогласия пользователя**: только логировать / использовать как validation set
* **Метрики сравнения старой и новой модели**: F1, accuracy — порог для деплоя
